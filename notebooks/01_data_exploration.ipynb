# 01 - Data Exploration

Objectif : comprendre le fichier brut, les colonnes disponibles et la qualite des donnees.

```python
import pandas as pd
from pathlib import Path

project_root = Path.cwd()
if project_root.name == 'notebooks':
    project_root = project_root.parent

raw_path = project_root / 'data' / 'raw' / 'Building_Energy_Benchmarking_Data,_2015-Present_20260704.csv'
```

## Charger les donnees brutes

```python
df = pd.read_csv(raw_path, low_memory=False)
print(f'Lignes : {df.shape[0]:,}')
print(f'Colonnes : {df.shape[1]}')
df.head()
```

## Apercu des colonnes

```python
pd.DataFrame({
    'column_name': df.columns,
    'dtype': df.dtypes.astype(str),
    'missing_values': df.isna().sum().values,
    'missing_rate': (df.isna().mean().values * 100).round(2)
}).sort_values('missing_rate', ascending=False).head(20)
```

## Doublons et annees disponibles

```python
print('Doublons complets :', df.duplicated().sum())
if 'DataYear' in df.columns:
    display(df['DataYear'].value_counts().sort_index())
```

## Colonnes utiles pour la suite

```python
useful_columns = [
    'OSEBuildingID', 'DataYear', 'BuildingName', 'BuildingType', 'Neighborhood',
    'YearBuilt', 'NumberofFloors', 'PropertyGFATotal', 'ENERGYSTARScore',
    'SiteEUI(kBtu/sf)', 'SiteEnergyUse(kBtu)', 'Electricity(kWh)',
    'NaturalGas(therms)', 'TotalGHGEmissions', 'GHGEmissionsIntensity',
    'ComplianceStatus'
]

[col for col in useful_columns if col in df.columns]
```
